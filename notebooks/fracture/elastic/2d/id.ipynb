```julia
using Plots
```

```julia
# phase field Gaussian*step_function

Nx = 40
Ny = 40

x = range(0,1,Nx)
y = range(0,1,Ny)

d0 = zeros(Nx, Ny)
f0 = 0.5*(1.0 .+ tanh.(50*(y .- 0.5)))
g0 = 0.95*exp.(-(x .- 0.5).^2/(2*0.00125))
for j in 1:Ny
    for i in 1:Nx
        #d0[i,j] = g0[i]*f0[j]
        if i == Int(Nx/2)
            d0[i,j] = f0[j]
        end 
    end
end
```

```julia
# transpose
dp = d0'

# coordinate matrices
X = [xi for yi in y, xi in x]
Y = [yi for yi in y, xi in x]

plt = plot(
    aspect_ratio = 1,
    xlabel = "x",
    ylabel = "y",
    legend = false,
    title = "Elastic Square"
)

# node heatmap
scatter!(
    plt,
    vec(X),
    vec(Y),
    marker = :square,
    marker_z = vec(dp),
    ms = 6,
    color = :coolwarm,
    markerstrokewidth = 0,   # removes black circle border
    colorbar = true,
    clim=(0,1)
)

# grid lines
for i in 1:Ny
    plot!(plt, X[i,:], Y[i,:], color=:black, lw=1)
end

for j in 1:Nx
    plot!(plt, X[:,j], Y[:,j], color=:black, lw=1)
end
```
