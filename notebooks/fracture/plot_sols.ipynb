```julia
using Serialization
using Plots

# ----- Load saved data -----
(x, u_all, d_all, H_all) = deserialize("../../examples/fracture/fracture.jls")
```

```julia
# Phase field evolution
plot(title="Phase Field Evolution", xlabel="x", ylabel="d", legend=:topright)
for i in 1:length(d_all)
    plot!(x, d_all[i], label="Iter $i")
end
display(current())
```

```julia
# Displacement evolution
plot(title="Displacement Evolution", xlabel="x", ylabel="u", legend=:topright)
for i in 1:length(u_all)
    plot!(x, u_all[i], label="Iter $i")
end
display(current())
```

```julia
# History field evolution
plot(title="History Field Evolution", xlabel="x", ylabel="H", legend=:topright)
for i in 1:length(H_all)
    plot!(x, H_all[i], label="Iter $i")
end
display(current())
```

```julia
# Gaussian bump (to seed damage / history field)
function gaussian(x, x0, width, amp)
    return amp * exp.(-(x.-x0).^2 ./ (2*width^2))
end
f = gaussian(x, 0.5, 0.05, 0.01)

plot(x,f)
```
